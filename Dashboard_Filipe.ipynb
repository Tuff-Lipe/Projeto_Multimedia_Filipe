```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt
# importamos a class wordcloud para gerar uma nuvem de palavras e stopwords para escolher o número de palavras
texto = "Na disciplina de Multimédia vamos aprender algumas coisas interessantes."
as_minhas_stopwords = set(['na', 'de', 'vamos', 'algumas', '.']) # Conjunto de palavras removidas da nuvem

# ---------------- Configurar e gerar a nuvem de palavras a partir do texto ----------------

nuvem = WordCloud(
    width = 800,
    height = 600,
    background_color = 'white',
    colormap = 'viridis', # 'plasma', 'Blues', 'Reds', 'Greens'
    max_words = 8,
    stopwords = as_minhas_stopwords
).generate(texto)

# ---------------- Exibir a imagem através do matplotlib ----------------

plt.figure(figsize = (10, 5)) # Define o tamanho da figura
plt.imshow(nuvem, interpolation = 'bilinear') # Renderiza a imagem
plt.axis('off') # Remove os eixos x e y
plt.show() # Exibe a imagem
```

# Programa 2 - Jogo da Adivinha

```python
print('---------------------- BEM VINDO AO JOGO DA ADIVINHA ----------------------')

import random
numero_secreto = random.randrange(1, 101)
total_tentativas = 0

print('Escolha o nível de dificuldade:')
print('1 - Fácil')
print('2 - Médio')
print('3 - Difícil')

nivel = int(input('Digite o número correspondente ao nível de dificuldade:'))

if nivel == 1:
    total_tentativas = 30
elif nivel == 2:
    total_tentativas = 20
elif nivel == 3:
    total_tentativas = 10
else:
    print('Nível inválido')

for jogada in range(1, total_tentativas + 1):
    print(f'Tentativa {jogada} de {total_tentativas}')
    valor_escolhido = int(input('Digite um número entre 1 e 100: '))
    if valor_escolhido < 1 or valor_escolhido > 100:
        print('Deve digitar um número entre 1 e 100!')
        continue
    if valor_escolhido == numero_secreto:
        print(f'Parabéns! Acertou o número secreto {numero_secreto} em {jogada} tentativas!')
        break
    elif valor_escolhido < numero_secreto:
        print(f'O número secreto é maior do que {valor_escolhido}')
    else:
        print(f'O número secreto é menor do que {valor_escolhido}')
    if jogada == total_tentativas:
        print(f'Não conseguiu adivinhar o número secreto {numero_secreto}. Tente novamente!')
```

# Programa 3 - Quiz das Capitais

```python
print('---------------------- BEM VINDO AO QUIZ ----------------------')

answer = input('Vamos jogar? (S/N)')
score = 0
total_questions = 5

if answer.upper() == 'S':
    answer = input('Pergunta 1: Qual é a capital de Portugal?')
    if answer.lower() == 'lisboa':
        score += 1
        print('CERTO!')
    else:
        print('Resposta errada...') 
   
    answer = input('Pergunta 2: Qual é a capital da Austrália?')
    if answer.lower() == 'canberra':
        score += 1
        print('CERTO!')
    else:
        print('Resposta errada...')
    
    answer = input('Pergunta 3: Qual é a capital da Suécia?')
    if answer.lower() == 'estocolmo':
        score += 1
        print('CERTO!')
    else:
         print('Resposta errada...')

    answer = input('Pergunta 3: Qual é a capital da Albânia?')
    if answer.lower() == 'tirana':
        score += 1
        print('CERTO!')
    else:
         print('Resposta errada...')

    answer = input('Pergunta 3: Qual é a capital do Chile?')
    if answer.lower() == 'santiago':
        score += 1
        print('CERTO!')
    else:
         print('Resposta errada...')

    pontuacao = int(round((score/total_questions)*100))
    print(f'Obrigado por jogares este Quiz! A tua pontuação foi {pontuacao}%!')
elif answer.upper == 'N':
    print('ok...')
else:
    print('Resposta inválida')
```
